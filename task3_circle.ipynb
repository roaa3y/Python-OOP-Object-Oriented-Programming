```python
class Calculator:

    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divideInt(self, a, b):
        return a // b

    def divideFloat(self, a, b):
        return a / b

    def modulus(self, a, b):
        return a % b


calc = Calculator()

print("add =", calc.add(15, 10))
print("subtract=", calc.subtract(20, 8))
print("multiply", calc.multiply(6, 7))
print("divideint", calc.divideInt(17, 5))
print("dividefloat", calc.divideFloat(17, 5))
print("modulus", calc.modulus(17, 5))
```

```python
class Circle:


    def __init__(self, radius,pi=3.14):
        self.radius = radius
        self.pi = pi

    def area(self):
        area = self.pi * (self.radius ** 2)
        return area


c1 = Circle(5)

print("Radius =", c1.radius)
print("Area =", c1.area())
```
