```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)


class Student(Person):
    pass


class std(Person):
    pass


class roaa(Person):
    pass


class shahed(Person):
    pass


y = std("CHILD 2", "INSIDE PRINT")
y.printname()

x = Student("Mike", "Olsen")
x.printname()

r = roaa("Roaa", "Ahmad")
r.printname()

s = shahed("Shahed", "Ali")
s.printname()
```
