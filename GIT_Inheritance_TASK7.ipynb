```python
class Parent:
    def __init__(self, fname, lastname, birthdate):
        self.fname = fname
        self.lastname = lastname
        self.birthdate = birthdate

    def print(self):
        print("First Name:", self.fname)
        print("Last Name:", self.lastname)
        print("Birthdate:", self.birthdate)


class Child1(Parent):
    pass


class Child2(Parent):
    def __init__(self, fname, lastname, birthdate, father_name):
        super().__init__(fname, lastname, birthdate)
        self.father_name = father_name

    def print(self):
        super().print()
        print("Father Name:", self.father_name)


class Child3(Parent):
    def __init__(self, fname, lastname, birthdate, year):
        super().__init__(fname, lastname, birthdate)
        self.year = year

    def print(self):
        super().print()
        print("Year:", self.year)


obj1 = Child1("Roaa", "Yaser", "25/11/2006")
obj2 = Child2("Roaa", "Yaser", "28/03/2005", "Ahmad")
obj3 = Child3("Roaa", "Yaser", "22/04/2006", 2026)

obj1.print()
print()

obj2.print()
print()

obj3.print()
```
